```python
class Produto:
    def __init__(self,nome, preco,quantidade):
        self.nome = nome
        self.preco = preco
        self.quantidade = quantidade
    def valor_total(self):
        return self.preco * self.quantidade
#metodo adicionar quantidade ao estoque
    def adicionar_estoque(self, adicionar):
        if adicionar > 0:
            self.quantidade += adicionar
            print(f"estoque aumentou em {adicionar} unid")
        else:
            print("valor invalido")
    def __str__(self):           # facilita print(produto)
        return f"{self.nome} | R${self.preco:.2f} x {self.quantidade} = R${self.valor_total():.2f}"
```

```python
p = Produto("Caneta", 2.50, 100)
print(p.valor_total())    # 250.0
p.adicionar_estoque(20)   # Estoque aumentado em 20 unid.
print(p.quantidade)       # 120
p.adicionar_estoque(-5)   # Valor inválido
```

```python
p = Produto("Arroz", 5.50, 10)
p.adicionar_estoque(5)
print(p) 
```

```python
#lista de  produto perecivel
from datetime import date
class Produtoperecivel(Produto):
    def __init__(self,nome, preco, quantidade, data_validade):
        super().__init__(nome, preco, quantidade)
        self.data_validade=data_validade
    #produto vencido
    def esta_vencido(self):
        return date.today()> self.data_validade
    #produto normal e vencido
    def __str__(self):
        status = "VENCIDO" if self.esta_vencido() else "válido"
        return f"{super().__str__()} | Validade: {self.data_validade} ({status})"
```

```python
leite = Produtoperecivel("Leite", 4.20, 20, date(2026, 11, 1))
leite.adicionar_estoque(10)          # herdado!
print(leite.valor_total())           # herdado!
```
